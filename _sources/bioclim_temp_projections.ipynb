# Projected bioclimatic variables - temperature

```python
import hvplot
import pandas as pd
import cartopy.crs as ccrs
import hvplot.xarray
import hvplot.pandas
import xarray as xr
from bokeh.plotting import figure, show, output_notebook
import bokeh

# output_notebook()

# https://github.com/the-turing-way/the-turing-way/blob/main/book/website/_toc.yml

# copy file structure from book above to reduce html sizes!

tiles_state=True

xlims=(10475351,15807367)
ylims=(-1200000,2630000)
```

This page displays the worldclim elevation and bioclimatic data variables. The contents menu at the top right can be used to navigate the page. Information on the variables can be found at https://www.worldclim.org/data/bioclim.html.

## BIO1: Annual air mean temperature (°C)

```python
ds = xr.open_dataarray('projected_bioclim/ensemble_mean_rcp85_ba_month_x++_q100_bio_1.tif')

ds = ds.sel(band=1).rename('Annual mean air temperature')-273.15
ds = ds.sel(x=slice(45,180),y=slice(-60,60))

hover = bokeh.models.HoverTool(tooltips=[("Temp", "@image{1.1f} °C"),
                                  ],
                              formatters = {"@Temp":"printf"})

ds.drop(['band','spatial_ref']).hvplot(x='x',y='y',geo=True,xlim=(95,142), ylim=(-13,25),width=700,height=450,cmap='magma',xlabel='lon',tools=[hover],tiles=tiles_state,tiles_opts={'xlim':xlims,'ylim':ylims,'alpha':0.5})
```
