# *Leptocybe invasa* case study

```python
import pandas as pd
import cartopy.crs as ccrs
import hvplot.xarray
import hvplot.pandas
import rioxarray as rx
from bokeh.plotting import figure, show, output_notebook
import bokeh
import numpy as np
from holoviews.util.transform import lon_lat_to_easting_northing
import rasterio
import xarray as xr
```

This page contains the results of a case study into *Leptocybe invasa*, using the MaxEnt species distribution modelling technique. Details on the settings used in the modelling can be found in the description at the bottom of the page.

```python

tiles_state=True

xlims=(10475351,15807367)
ylims=(-1200000,2630000)

# opens the elevation data as an xarray
ds = xr.open_dataarray('worldclim_data/wc2.1_10m_elev.tif')

# opens the species data as a pandas dataframe
df = pd.read_csv('species_data/leptocybe-invasa_aggregated.csv')

# selects the 1st band of the tif and renames it to the relevant variable
ds = ds.sel(band=1).rename('Elevation')

# sets the information displayed by the hovertool in the image, including the unit displayed after the number (in this case m).
hover = bokeh.models.HoverTool(tooltips=[("Elevation", "@image m"),
                                  ],
                              formatters = {"@Elevation":"printf"})

# plots the image
ele = ds.drop_vars(['band','spatial_ref']).hvplot(x='x',y='y',geo=True,xlim=(89,147), ylim=(-15,28),width=800,height=500,cmap='plasma',xlabel='Longitude',tools=[hover],tiles=tiles_state)

locs = df.hvplot.points('lon', 'lat', tiles=True, color='red',alpha=0.8,geo=True,xlim=(89,147),ylim=(-15,28),width=800,height=500)
```

### Historic relative occurrence of probability

```python
ds = xr.open_dataarray('rop_data/hist_summary_rop.tif')
ds = ds.sel(band=1).rename('Historic ROP')
ds = ds.sel(x=slice(45,180),y=slice(-60,60))

hover = bokeh.models.HoverTool(tooltips=[("ROP", "@image{1.3f}"),
                                  ],
                              formatters = {"@ROP":"printf"})

ds.drop(['band','spatial_ref']).hvplot(x='x',y='y',geo=True,xlim=(95,142), ylim=(-13,25),width=700,height=450,cmap='plasma',xlabel='lon',tools=[hover],tiles=tiles_state,tiles_opts={'xlim':xlims,'ylim':ylims,'alpha':0.5})
```

### Future relative occurrence of probability

```python
ds = xr.open_dataarray('rop_data/fut_summary_rop.tif')
ds = ds.sel(band=1).rename('Future ROP')
ds = ds.sel(x=slice(45,180),y=slice(-60,60))

hover = bokeh.models.HoverTool(tooltips=[("ROP", "@image{1.3f}"),
                                  ],
                              formatters = {"@ROP":"printf"})

ds.drop(['band','spatial_ref']).hvplot(x='x',y='y',geo=True,xlim=(95,142), ylim=(-13,25),width=700,height=450,cmap='plasma',xlabel='lon',tools=[hover],tiles=tiles_state,tiles_opts={'xlim':xlims,'ylim':ylims,'alpha':0.5})
```

### Change in relative occurrence of probability

```python
ds = xr.open_dataarray('rop_data/rop_change.tif')
ds = ds.sel(band=1).rename('Change in ROP')
ds = ds.sel(x=slice(45,180),y=slice(-60,60))

hover = bokeh.models.HoverTool(tooltips=[("ΔROP", "@image{1.3f}"),
                                  ],
                              formatters = {"@ΔROP":"printf"})

ds.drop(['band','spatial_ref']).hvplot(x='x',y='y',clim=(-0.60463027,0.60463027),geo=True,xlim=(95,142), ylim=(-13,25),width=700,height=450,cmap='bwr',xlabel='lon',tools=[hover],tiles=tiles_state,tiles_opts={'xlim':xlims,'ylim':ylims,'alpha':0.5})
```
